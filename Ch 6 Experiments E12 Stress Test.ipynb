## Additional Plots 

### A. Growth in model size

```python

import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# --------------------------------------------------
# Formatting: LaTeX-style font
# --------------------------------------------------

plt.rcParams["font.family"] = "serif"


# --------------------------------------------------
# Hard-coded E12 model sizes
# --------------------------------------------------

# Equal-sized synthetic networks
network_size = [30, 40, 50, 60, 70, 80]

# Number of decision variables
variables = [
    576210,
    1024280,
    1600350,
    2304420,
    3136490,
    4096560
]


# --------------------------------------------------
# Create figure
# --------------------------------------------------

fig, ax = plt.subplots(figsize=(10, 7))


# --------------------------------------------------
# Plot model size
# --------------------------------------------------

ax.plot(
    network_size,
    variables,
    color="black",
    marker="o",
    markersize=5,
    linewidth=1,
    linestyle="-"
)


# --------------------------------------------------
# Axis formatting
# --------------------------------------------------

ax.set_xlabel("Suppliers, candidate plants and buyers per network")
ax.set_ylabel("Number of decision variables")

ax.set_xticks(network_size)

ax.yaxis.set_major_formatter(
    FuncFormatter(
        lambda x, pos: f"{x / 1e6:.1f}"
    )
)

# Indicate units in millions
ax.set_ylabel("Decision variables (million)")


# --------------------------------------------------
# Formatting
# --------------------------------------------------

ax.grid(
    True,
    linestyle=":",
    linewidth=0.5,
    alpha=0.5
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.set_ylim(bottom=0)

plt.tight_layout()
plt.show()
```

### B. Solver time versus network size

```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import os

# --------------------------------------------------
# E12 – Hard-coded results
# --------------------------------------------------

scenarios = [
    "E12.0", "E12.1", "E12.2", "E12.3",
    "E12.4", "E12.5", "E12.6"
]

network_sizes = [
    "20/15/18", "30/30/30", "40/40/40",
    "50/50/50", "60/60/60", "70/70/70",
    "80/80/80"
]

x = np.arange(len(scenarios))

# Solver times (seconds)
cost_times = np.array([
    20.53, 1176.14, 3609.03, 3617.47,
    3620.92, 3648.55, 3612.83
])

emissions_times = np.array([
    2.41, 11.58, 71.01, 272.84,
    69.45, 535.80, 687.09
])

# True = required MIP gap reached
cost_solved = np.array([
    True, True, False, False, False, False, False
])

emissions_solved = np.array([
    True, True, True, True, True, True, True
])

time_limit = 3600

# --------------------------------------------------
# Figure formatting
# --------------------------------------------------

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 11,
    "axes.labelsize": 12,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 10
})

fig, ax = plt.subplots(figsize=(10, 5.5))

# --------------------------------------------------
# Min-cost
# --------------------------------------------------

# Connect only successful runs
ax.plot(
    x[:2], cost_times[:2],
    color="black",
    linestyle="-",
    linewidth=1.6,
    marker="o",
    markersize=7,
    label="Min-cost"
)

# Time-limit runs: hollow circles, not connected
ax.scatter(
    x[~cost_solved],
    cost_times[~cost_solved],
    marker="o",
    s=65,
    facecolors="white",
    edgecolors="black",
    linewidths=1.5,
    zorder=5
)

# --------------------------------------------------
# Min-emissions
# --------------------------------------------------

ax.plot(
    x, emissions_times,
    color="black",
    linestyle="--",
    linewidth=1.6,
    marker="s",
    markersize=6,
    label="Min-emissions"
)

# --------------------------------------------------
# Time-limit reference
# --------------------------------------------------

ax.axhline(
    y=time_limit,
    color="grey",
    linestyle=":",
    linewidth=1.2
)

# --------------------------------------------------
# Axes and formatting
# --------------------------------------------------

ax.set_xlabel("Network size (suppliers/plants/buyers)")
ax.set_ylabel("Solver time (s)")

ax.set_xticks(x)
ax.set_xticklabels(network_sizes)

ax.set_ylim(0, 3950)
ax.set_xlim(-0.35, len(x) - 0.65)

ax.grid(
    axis="y",
    linestyle=":",
    linewidth=0.6,
    alpha=0.5
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# --------------------------------------------------
# Legend – Outside on the right
# --------------------------------------------------

legend_handles = [
    Line2D(
        [0], [0],
        color="black",
        linestyle="-",
        marker="o",
        markersize=7,
        label="Min-cost"
    ),
    Line2D(
        [0], [0],
        color="black",
        linestyle="--",
        marker="s",
        markersize=6,
        label="Min-emissions"
    ),
    Line2D(
        [0], [0],
        color="black",
        linestyle="None",
        marker="o",
        markerfacecolor="white",
        markeredgecolor="black",
        markersize=7,
        label="Time limit exceeded"
    ),
    Line2D(
        [0], [0],
        color="grey",
        linestyle=":",
        linewidth=1.2,
        label="3,600 s time limit"
    )
]

ax.legend(
    handles=legend_handles,
    loc="upper left",
    bbox_to_anchor=(1.02, 1),
    frameon=False,
    borderaxespad=0
)

plt.tight_layout()

# --------------------------------------------------
# Save and display
# --------------------------------------------------

filename = "e12_solver_time.png"

plt.savefig(
    filename,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Figure saved to:", os.path.abspath(filename))
```
